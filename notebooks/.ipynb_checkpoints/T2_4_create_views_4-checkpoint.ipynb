---
### Why the DBRepo mapper requires a specific column strategy

DBRepo's internal `query_to_subset` mapper (line 32–34 of `mapper.py`) does this:
```python
array([table.columns for table in database.tables if table.internal_name in query_table_names])
```
It collects each involved table's full column list and calls `numpy.array()` on them.
If tables have **different numbers of columns**, numpy raises `ValueError: inhomogeneous shape`.

**The fix:** include ALL columns from the primary table (in `table.column` format),
but only **exactly ONE column per joined lookup table**. This keeps the inner arrays
uniform in size (each joined table contributes 1 item), which numpy can handle.

## 4. Fetch Full Table Metadata from DBRepo

`get_tables()` returns lightweight `TableBrief` objects with no column info.
We must call `get_table()` individually per table to get full column details.

````python
# Step 1: get brief list of all tables (id + name only)
all_tables_brief = client.get_tables(database_id=DATABASE_ID)

# Step 2: fetch full table object (includes columns) for each
table_by_name = {}
for t in all_tables_brief:
    full_table = client.get_table(database_id=DATABASE_ID, table_id=t.id)
    table_by_name[full_table.name] = full_table

# Print all tables and their columns for verification
print("Tables in database:")
for name, tbl in table_by_name.items():
    col_names = [c.internal_name for c in tbl.columns]
    print(f"  {name} ({len(col_names)} cols): {col_names}")
````

## 5. Helper Functions

````python
def get_primary_columns(table_by_name, table_name, exclude=None):
    """
    Returns a list of 'table.column' strings for the PRIMARY table in a view.
    All columns are included except those listed in `exclude`.

    Args:
        table_by_name: dict of {table_name: full_table_object}
        table_name:    name of the primary (FROM) table
        exclude:       list of column names to omit (FK ids, duplicates, etc.)

    Returns:
        List of 'table.column' strings
    """
    exclude = exclude or []
    tbl = table_by_name[table_name]
    return [
        f"{table_name}.{col.internal_name}"
        for col in tbl.columns
        if col.internal_name not in exclude
    ]


def create_view_safe(client, database_id, name, query):
    """
    Creates a view only if one with that name does not already exist.
    Safe to re-run — skips creation if the view is already present.

    Args:
        client:      authenticated RestClient
        database_id: UUID of the target database
        name:        view name
        query:       QueryDefinition object

    Returns:
        view id (string)
    """
    existing = {v.name: v.id for v in client.get_views(database_id=database_id)}
    if name in existing:
        print(f"[SKIP] '{name}' already exists (id: {existing[name]})")
        return existing[name]

    view = client.create_view(
        database_id=database_id,
        name=name,
        query=query,
        is_public=True,
        is_schema_public=True,
    )
    print(f"[OK]   '{name}' created successfully (id: {view.id})")
    return view
````

## 6. View 1 — `vw_forward_features`

**Purpose:** Joins `forward_player_valuation` with `player` and `club` lookup tables.
One flat row per player with all performance features and the ML target variable.

**Source:** Dataset 1 — *Forward football player valuation* (Briseño & Rivera, 2024)  
**Target variable:** `market_value_mln`  
**Used in:** T2.6 ML pipeline data loading

**Column strategy:**
- Primary table `forward_player_valuation`: all meaningful columns (excluding internal FK ids)
- Joined table `player`: exactly `player.player_name` (1 column)
- Joined table `club`: exactly `club.club_name` (1 column)

**Equivalent SQL:**
```sql
SELECT
    f.forward_valuation_id, f.player_age_years, f.plays_in_europe,
    f.matches_played, f.goals, f.assists, f.minutes_per_goal,
    f.minutes_played, f.instagram_followers_mln, f.value_rank,
    f.market_value_mln,   -- TARGET
    p.player_name,
    c.club_name
FROM forward_player_valuation f
INNER JOIN player p ON f.player_id = p.player_id
INNER JOIN club   c ON f.club_id   = c.club_id;
```

````python
# Primary table: all columns except internal FK ids and original-name duplicates
forward_primary_cols = get_primary_columns(
    table_by_name,
    table_name="forward_player_valuation",
    exclude=[
        "player_id",            # resolved via JOIN → player.player_name
        "club_id",              # resolved via JOIN → club.club_name
        "source_dataset_id",    # internal provenance, not an ML feature
        "original_player_name", # duplicate of player.player_name
        "original_team_name",   # duplicate of club.club_name
    ],
)

# Joined tables: exactly ONE column each to satisfy the numpy array constraint
# in DBRepo's mapper (all joined table column lists must be the same length)
forward_cols = forward_primary_cols + [
    "player.player_name",   # 1 col from player
    "club.club_name",       # 1 col from club
]

print("Columns for vw_forward_features:")
for c in forward_cols:
    print(f"  {c}")

query_forward = QueryDefinition(
    datasources=["forward_player_valuation"],
    columns=forward_cols,
    joins=[
        # Bring in player_name from the player lookup table
        JoinDefinition(
            type=JoinType.INNER,
            datasource="player",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.player_id",
                    foreign_column="player.player_id",
                )
            ],
        ),
        # Bring in club_name from the club lookup table
        JoinDefinition(
            type=JoinType.INNER,
            datasource="club",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.club_id",
                    foreign_column="club.club_id",
                )
            ],
        ),
    ],
)

view_forward = create_view_safe(
    client=client,
    database_id=DATABASE_ID,
    name=VIEW_FORWARD,
    query=query_forward,
)
````

## 7. View 2 — `vw_transfer_features`

**Purpose:** Joins `transfer_value_observation` with `player`, `position`, `nationality`,
and `club` lookup tables. One flat row per observation across all seasons.

**Source:** Dataset 2 — *Transfer Value Determinants* (Nisanov, 2025)  
**Target variable:** `value_end_mln`  
**Used in:** T2.6 ML pipeline data loading

**Column strategy:**
- Primary table `transfer_value_observation`: all meaningful columns
- Each of the 4 joined tables: exactly 1 column (the name field)

**Equivalent SQL:**
```sql
SELECT
    t.transfer_observation_id, t.season_year, t.age_then_years, t.height_cm,
    t.total_games, t.total_goals, t.assists, t.penalty_kicks, t.total_minutes,
    t.club_performance, t.relegation, t.success_or_not,
    t.value_start_mln, t.value_delta_mln, t.value_end_mln,  -- TARGET
    p.player_name, pos.position_name, n.nationality_name, c.club_name
FROM transfer_value_observation t
INNER JOIN player      p   ON t.player_id      = p.player_id
INNER JOIN position    pos ON t.position_id    = pos.position_id
INNER JOIN nationality n   ON t.nationality_id = n.nationality_id
INNER JOIN club        c   ON t.club_id        = c.club_id;
```

````python
# Primary table: all columns except FK ids, name duplicates, and raw EUR columns
transfer_primary_cols = get_primary_columns(
    table_by_name,
    table_name="transfer_value_observation",
    exclude=[
        "player_id", "position_id", "nationality_id", "club_id",  # resolved via JOINs
        "source_dataset_id",                                        # internal provenance
        "original_player_name", "original_position_name",          # duplicates
        "original_nationality_name", "original_club_name",         # duplicates
        "start_value_eur", "end_value_eur", "delta_value_eur",     # use _mln versions
        "age_now_years",    # current age less informative for ML than age at transfer
    ],
)

# Joined tables: exactly ONE column each
# All four joined tables contribute 1 column → uniform array size for numpy
transfer_cols = transfer_primary_cols + [
    "player.player_name",           # 1 col from player
    "position.position_name",       # 1 col from position
    "nationality.nationality_name", # 1 col from nationality
    "club.club_name",               # 1 col from club
]

print("Columns for vw_transfer_features:")
for c in transfer_cols:
    print(f"  {c}")

query_transfer = QueryDefinition(
    datasources=["transfer_value_observation"],
    columns=transfer_cols,
    joins=[
        # player_name
        JoinDefinition(
            type=JoinType.INNER,
            datasource="player",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.player_id",
                    foreign_column="player.player_id",
                )
            ],
        ),
        # position_name
        JoinDefinition(
            type=JoinType.INNER,
            datasource="position",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.position_id",
                    foreign_column="position.position_id",
                )
            ],
        ),
        # nationality_name
        JoinDefinition(
            type=JoinType.INNER,
            datasource="nationality",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.nationality_id",
                    foreign_column="nationality.nationality_id",
                )
            ],
        ),
        # club_name
        JoinDefinition(
            type=JoinType.INNER,
            datasource="club",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.club_id",
                    foreign_column="club.club_id",
                )
            ],
        ),
    ],
)

view_transfer = create_view_safe(
    client=client,
    database_id=DATABASE_ID,
    name=VIEW_TRANSFER,
    query=query_transfer,
)
````

## 8. View 3 — `vw_combined_player_value`

**Purpose:** Shared-schema view over Dataset 1 exposing only the columns that have
a direct equivalent in both datasets. Useful for cross-dataset exploratory analysis.

**Column strategy:**
- Primary table: only the shared feature columns (age, goals, assists, minutes, value)
- Joined tables: exactly 1 column each (`player.player_name`, `club.club_name`)

> **Note:** DBRepo does not support `UNION ALL` in `QueryDefinition`.
> The full UNION SQL is saved in `/data/views.sql` in the repository for reference.

**Equivalent SQL (full UNION — reference only, see `/data/views.sql`):**
```sql
SELECT p.player_name, c.club_name,
       f.player_age_years AS age_years, f.goals, f.assists,
       f.minutes_played, f.market_value_mln, 'dataset_1' AS source
FROM forward_player_valuation f
INNER JOIN player p ON f.player_id = p.player_id
INNER JOIN club   c ON f.club_id   = c.club_id
UNION ALL
SELECT p.player_name, c.club_name,
       t.age_then_years, t.total_goals, t.assists,
       t.total_minutes, t.value_end_mln, 'dataset_2' AS source
FROM transfer_value_observation t
INNER JOIN player p ON t.player_id = p.player_id
INNER JOIN club   c ON t.club_id   = c.club_id;
```

````python
# Primary table: only the columns shared across both datasets
# Excluded: anything with no equivalent in Dataset 2
combined_primary_cols = get_primary_columns(
    table_by_name,
    table_name="forward_player_valuation",
    exclude=[
        "forward_valuation_id",     # internal id
        "source_dataset_id",        # internal provenance
        "player_id",                # resolved via JOIN
        "club_id",                  # resolved via JOIN
        "original_player_name",     # duplicate
        "original_team_name",       # duplicate
        "plays_in_europe",          # no equivalent in Dataset 2
        "matches_played",           # Dataset 2 uses total_games instead
        "minutes_per_goal",         # derived stat, no equivalent in Dataset 2
        "instagram_followers_mln",  # no equivalent in Dataset 2
        "value_rank",               # no equivalent in Dataset 2
        # Kept: player_age_years, goals, assists, minutes_played,
        #       market_value_mln (TARGET)
    ],
)

# Joined tables: exactly 1 column each
combined_cols = combined_primary_cols + [
    "player.player_name",   # 1 col from player
    "club.club_name",       # 1 col from club
]

print("Columns for vw_combined_player_value:")
for c in combined_cols:
    print(f"  {c}")

query_combined = QueryDefinition(
    datasources=["forward_player_valuation"],
    columns=combined_cols,
    joins=[
        JoinDefinition(
            type=JoinType.INNER,
            datasource="player",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.player_id",
                    foreign_column="player.player_id",
                )
            ],
        ),
        JoinDefinition(
            type=JoinType.INNER,
            datasource="club",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.club_id",
                    foreign_column="club.club_id",
                )
            ],
        ),
    ],
)

view_combined = create_view_safe(
    client=client,
    database_id=DATABASE_ID,
    name=VIEW_COMBINED,
    query=query_combined,
)
````

## 9. Verify — List All Views

````python
all_views   = client.get_views(database_id=DATABASE_ID)
view_id_map = {v.name: v.id for v in all_views}

print(f"Views registered in database '{DATABASE_ID}':")
print("-" * 60)
for v in all_views:
    print(f"  Name : {v.name}")
    print(f"  ID   : {v.id}")
    print()

missing = {VIEW_FORWARD, VIEW_TRANSFER, VIEW_COMBINED} - set(view_id_map)
if missing:
    print(f"WARNING: views not found: {missing}")
else:
    print("All three views registered successfully.")
````
