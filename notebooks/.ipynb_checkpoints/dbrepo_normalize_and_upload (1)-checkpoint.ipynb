```
source_dataset → player → club → position → nationality → season
    ↓                                                        ↓
forward_player_valuation              transfer_value_observation
```

## 5. Load Raw Data

````python
# ── Dataset 1: Forward player valuation ───────────────────────────────────────
forward_raw = pd.read_excel(FORWARD_FILE)
# Normalise column names to lowercase for consistent access
forward_raw.columns = [c.strip().lower() for c in forward_raw.columns]
print(f"Dataset 1 shape: {forward_raw.shape}")
print(f"Columns: {forward_raw.columns.tolist()}")

print()

# ── Dataset 2: Transfer Value Determinants (5 sheets: 2019–2023) ──────────────
# Each sheet represents one season. Concatenate all sheets.
# Sheets 2020–2023 lack club_performance/relegation/success_or_not — these
# become NaN after concat, which is correct (nullable in schema).
transfer_sheets = pd.read_excel(TRANSFER_FILE, sheet_name=None)
transfer_raw = pd.concat(
    [sheet.dropna(axis=1, how="all") for sheet in transfer_sheets.values()],
    ignore_index=True,
    sort=False,
)
transfer_raw.columns = [c.strip().lower() for c in transfer_raw.columns]
print(f"Dataset 2 shape: {transfer_raw.shape}")
print(f"Columns: {transfer_raw.columns.tolist()}")
````

## 6. Build Lookup Tables
These are the small reference tables that fact tables point to via FK.
Column names must exactly match the DBRepo schema.

````python
# ── player ────────────────────────────────────────────────────────────────────
# Combine player names from both datasets, deduplicate
player_names = pd.concat([
    forward_raw["nombre"].rename("player_name"),
    transfer_raw["player_name"],
]).dropna().drop_duplicates().sort_values().reset_index(drop=True)

player_df = pd.DataFrame({
    "player_id":   range(1, len(player_names) + 1),
    "player_name": player_names.values,
})
print(f"player: {len(player_df)} rows")

# ── club ──────────────────────────────────────────────────────────────────────
club_names = pd.concat([
    forward_raw["team"].rename("club_name"),
    transfer_raw["club_name"],
]).dropna().drop_duplicates().sort_values().reset_index(drop=True)

club_df = pd.DataFrame({
    "club_id":   range(1, len(club_names) + 1),
    "club_name": club_names.values,
})
print(f"club:   {len(club_df)} rows")

# ── position ──────────────────────────────────────────────────────────────────
# Schema column is 'position_name' — NOT 'position'
position_names = (
    transfer_raw["position"]
    .dropna().drop_duplicates().sort_values().reset_index(drop=True)
)
position_df = pd.DataFrame({
    "position_id":   range(1, len(position_names) + 1),
    "position_name": position_names.values,  # renamed to match schema
})
print(f"position: {len(position_df)} rows")

# ── nationality ───────────────────────────────────────────────────────────────
# Schema column is 'nationality_name' — NOT 'nationality'
nationality_names = (
    transfer_raw["nationality"]
    .dropna().drop_duplicates().sort_values().reset_index(drop=True)
)
nationality_df = pd.DataFrame({
    "nationality_id":   range(1, len(nationality_names) + 1),
    "nationality_name": nationality_names.values,  # renamed to match schema
})
print(f"nationality: {len(nationality_df)} rows")

# ── season ────────────────────────────────────────────────────────────────────
# The schema pre-inserts seasons 2019–2023 via SQL.
# We recreate that here for the upload. 'notes' matches the schema INSERT.
season_df = pd.DataFrame({
    "season_year": [2019, 2020, 2021, 2022, 2023],
    "notes": [
        "Season sheet in Transfer Value Determinants",
        "Season sheet in Transfer Value Determinants",
        "Season sheet in Transfer Value Determinants",
        "Season sheet in Transfer Value Determinants",
        "Season sheet in Transfer Value Determinants",
    ],
})
print(f"season: {len(season_df)} rows")
````

## 10. Schema Verification
Compare every DataFrame's columns against what DBRepo actually has registered.
All must show `MATCH: True` before uploading.

````python
def verify_schema(client, database_id, table_id_map, table_name, dataframe):
    """Compare DataFrame columns against DBRepo registered columns for a table."""
    table  = client.get_table(database_id=database_id, table_id=table_id_map[table_name])
    db_cols = [c.internal_name for c in table.columns]
    df_cols = dataframe.columns.tolist()
    match   = db_cols == df_cols
    status  = "OK" if match else "MISMATCH"
    print(f"[{status}] {table_name}")
    if not match:
        only_db = set(db_cols) - set(df_cols)
        only_df = set(df_cols) - set(db_cols)
        if only_db: print(f"       In DBRepo only : {only_db}")
        if only_df: print(f"       In DataFrame only: {only_df}")
    return match

tables_to_verify = [
    ("source_dataset",            source_dataset_df),
    ("player",                    player_df),
    ("club",                      club_df),
    ("position",                  position_df),
    ("nationality",               nationality_df),
    ("season",                    season_df),
    ("forward_player_valuation",  forward_fact),
    ("transfer_value_observation",transfer_fact),
]

all_ok = True
for name, df in tables_to_verify:
    ok = verify_schema(client, DATABASE_ID, table_id_map, name, df)
    all_ok = all_ok and ok

print()
if all_ok:
    print("All schemas match. Safe to upload.")
else:
    print("Fix mismatches above before uploading.")
````

## 12. Upload to DBRepo

Upload order strictly follows FK dependencies:
- Lookup tables first (`source_dataset`, `player`, `club`, `position`, `nationality`, `season`)
- Fact tables last (`forward_player_valuation`, `transfer_value_observation`)

The SDK's `import_table_data()` requires a **file path** — we write each
DataFrame to a temp CSV and pass that path, then clean up immediately after.

> **Note:** HTTP 204 from the server means success. The SDK may raise
> `ResponseCodeError: 204 is not 202` — this is a known SDK bug. The data
> was still uploaded. Check the DBRepo UI to confirm.

````python
def upload_table(client, database_id, table_id_map, table_name, dataframe):
    """
    Upload a DataFrame to a DBRepo table.

    The SDK requires a file path, not a DataFrame, so we write to a
    temp CSV file, upload it, then delete the temp file.

    HTTP 204 = success (SDK bug treats it as error — we catch and report correctly).
    """
    table_id = table_id_map.get(table_name)
    if not table_id:
        print(f"[ERROR] '{table_name}' not found in DBRepo table map. Skipping.")
        return False

    # Write to a temp CSV file
    tmp_path = None
    try:
        with tempfile.NamedTemporaryFile(
            mode="w", suffix=".csv", delete=False, encoding="utf-8"
        ) as tmp:
            dataframe.to_csv(tmp, index=False)
            tmp_path = tmp.name

        client.import_table_data(
            database_id=database_id,
            table_id=table_id,
            file_path=tmp_path,
            separator=",",
            skip_lines=1,        # skip the CSV header row
            line_encoding="\n",
        )
        print(f"[OK]    {table_name} ({len(dataframe)} rows)")
        return True

    except Exception as e:
        error_str = str(e)
        # HTTP 204 is actually a success — the SDK incorrectly rejects it
        if "204" in error_str and "202" in error_str:
            print(f"[OK*]   {table_name} ({len(dataframe)} rows) — HTTP 204 (success, SDK bug)")
            return True
        else:
            print(f"[FAIL]  {table_name}")
            print(f"        {type(e).__name__}: {error_str}")
            return False

    finally:
        # Always remove the temp file
        if tmp_path and os.path.exists(tmp_path):
            os.unlink(tmp_path)
````

````python
# Upload in FK dependency order — DO NOT reorder these
upload_order = [
    # ── Lookup tables (no FK dependencies) ──────────────────────────────────
    ("source_dataset",             source_dataset_df),
    ("player",                     player_df),
    ("club",                       club_df),
    ("position",                   position_df),
    ("nationality",                nationality_df),
    ("season",                     season_df),
    # ── Fact tables (depend on all lookup tables above) ──────────────────────
    ("forward_player_valuation",   forward_fact),
    ("transfer_value_observation", transfer_fact),
]

print("Starting upload...")
print("-" * 50)
results = {}
for table_name, dataframe in upload_order:
    results[table_name] = upload_table(
        client=client,
        database_id=DATABASE_ID,
        table_id_map=table_id_map,
        table_name=table_name,
        dataframe=dataframe,
    )

print("-" * 50)
passed = sum(results.values())
print(f"\nUpload complete: {passed}/{len(results)} tables succeeded.")
if passed < len(results):
    failed = [k for k, v in results.items() if not v]
    print(f"Failed tables: {failed}")
````

## 13. Post-Upload Verification
Query each table via the API to confirm rows were actually stored.
Run after upload completes.

````python
# Fetch full table objects to get column metadata
print("Row counts via DBRepo API:")
print("-" * 40)

for table_name, expected_df in upload_order:
    try:
        full_table = client.get_table(
            database_id=DATABASE_ID,
            table_id=table_id_map[table_name],
        )
        # get_table_data returns a DataFrame
        result_df = client.get_table_data(
            database_id=DATABASE_ID,
            table_id=table_id_map[table_name],
        )
        actual_rows   = len(result_df)
        expected_rows = len(expected_df)
        status = "OK" if actual_rows == expected_rows else "MISMATCH"
        print(f"  [{status}] {table_name}: {actual_rows} rows (expected {expected_rows})")
    except Exception as e:
        print(f"  [ERROR] {table_name}: {e}")
````
