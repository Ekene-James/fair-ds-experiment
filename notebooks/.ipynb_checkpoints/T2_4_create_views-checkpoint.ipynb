## 4. Look Up Table IDs
DBRepo assigns a UUID to each table when it is created. Rather than hardcoding these,
we fetch all tables in the database and build a lookup dictionary by name.

````python
# Fetch all tables in the database
all_tables = client.get_tables(database_id=DATABASE_ID)

# Build a name → id lookup dictionary
table_ids = {t.name: t.id for t in all_tables}

# Print all found tables for verification
print("Tables found in database:")
for name, tid in table_ids.items():
    print(f"  {name}: {tid}")

# Check that every expected table is present before we try to create views
missing = [t for t in TABLE_NAMES if t not in table_ids]
if missing:
    raise RuntimeError(
        f"The following expected tables are missing from the database: {missing}\n"
        "Make sure Student A has created the tables (T2.1) before running this notebook."
    )

print("\nAll expected tables are present. Ready to create views.")
````

## 5. Helper Function
A small utility that checks if a view with a given name already exists.
This prevents errors if the notebook is re-run.

````python
def get_existing_views(client, database_id):
    """
    Returns a dictionary of {view_name: view_id} for all views
    currently in the database.
    """
    existing = client.get_views(database_id=database_id)
    return {v.name: v.id for v in existing}


def create_view_safe(client, database_id, name, query):
    """
    Creates a view only if it does not already exist.
    If it exists, prints a message and skips creation.
    Returns the view object either way.
    """
    existing = get_existing_views(client, database_id)

    if name in existing:
        print(f"[SKIP] View '{name}' already exists (id: {existing[name]}). Skipping.")
        return existing[name]

    view = client.create_view(
        database_id=database_id,
        name=name,
        query=query,
        is_public=True,
        is_schema_public=True,
    )
    print(f"[OK]   View '{name}' created successfully (id: {view.id})")
    return view
````

## 6. View 1 — `vw_forward_features`

**Purpose:** Joins `forward_player_valuation` with the `player` and `club` lookup tables
into a single flat row per player. Exposes all performance and social media features
alongside the target variable `market_value_mln`.

**Source dataset:** Dataset 1 — *Forward football player valuation* (Briseño & Rivera, 2024)  
**Target variable (what the ML model predicts):** `market_value_mln`  
**Used in:** T2.6 data loading (ML pipeline, Dataset 1)

**Equivalent SQL:**
```sql
SELECT
    f.forward_valuation_id,
    p.player_name,
    c.club_name,
    f.player_age_years,
    f.plays_in_europe,
    f.matches_played,
    f.goals,
    f.assists,
    f.minutes_per_goal,
    f.minutes_played,
    f.instagram_followers_mln,
    f.value_rank,
    f.market_value_mln          -- TARGET VARIABLE
FROM forward_player_valuation f
INNER JOIN player p ON f.player_id = p.player_id
INNER JOIN club   c ON f.club_id   = c.club_id;
```

````python
# ── Define the query for vw_forward_features ──────────────────────────────────
#
# datasources: the primary (FROM) table
# columns:     every column we want in the view, prefixed with table name
# joins:       each JOIN clause — table to join, and which columns to match ON

query_forward = QueryDefinition(
    datasources=["forward_player_valuation"],
    columns=[
        # Identifier
        "forward_player_valuation.forward_valuation_id",
        # Player and club names (human-readable, from lookup tables)
        "player.player_name",
        "club.club_name",
        # Player features
        "forward_player_valuation.player_age_years",
        "forward_player_valuation.plays_in_europe",
        # Performance stats
        "forward_player_valuation.matches_played",
        "forward_player_valuation.goals",
        "forward_player_valuation.assists",
        "forward_player_valuation.minutes_per_goal",
        "forward_player_valuation.minutes_played",
        # Social media
        "forward_player_valuation.instagram_followers_mln",
        # Ranking
        "forward_player_valuation.value_rank",
        # TARGET: the market value the ML model is trained to predict
        "forward_player_valuation.market_value_mln",
    ],
    joins=[
        # Join player lookup table to get player_name instead of player_id
        JoinDefinition(
            type=JoinType.INNER,
            datasource="player",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.player_id",
                    foreign_column="player.player_id",
                )
            ],
        ),
        # Join club lookup table to get club_name instead of club_id
        JoinDefinition(
            type=JoinType.INNER,
            datasource="club",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.club_id",
                    foreign_column="club.club_id",
                )
            ],
        ),
    ],
)

# ── Create the view (skips if it already exists) ───────────────────────────────
view_forward = create_view_safe(
    client=client,
    database_id=DATABASE_ID,
    name=VIEW_FORWARD,
    query=query_forward,
)
````

## 7. View 2 — `vw_transfer_features`

**Purpose:** Joins `transfer_value_observation` with all five lookup tables (`player`, `position`,
`nationality`, `club`, `season`) into a flat row per observation. Exposes multi-season
transfer value data with all features and the target variable `value_end_mln`.

**Source dataset:** Dataset 2 — *Transfer Value Determinants* (Nisanov, 2025)  
**Target variable (what the ML model predicts):** `value_end_mln`  
**Used in:** T2.6 data loading (ML pipeline, Dataset 2)

**Equivalent SQL:**
```sql
SELECT
    t.transfer_observation_id,
    p.player_name,
    pos.position_name,
    n.nationality_name,
    c.club_name,
    t.season_year,
    t.age_then_years,
    t.height_cm,
    t.total_games,
    t.total_goals,
    t.assists,
    t.penalty_kicks,
    t.total_minutes,
    t.club_performance,
    t.relegation,
    t.success_or_not,
    t.value_start_mln,
    t.value_delta_mln,
    t.value_end_mln     -- TARGET VARIABLE
FROM transfer_value_observation t
INNER JOIN player      p   ON t.player_id      = p.player_id
INNER JOIN position    pos ON t.position_id    = pos.position_id
INNER JOIN nationality n   ON t.nationality_id = n.nationality_id
INNER JOIN club        c   ON t.club_id        = c.club_id;
-- Note: season_year is a direct column (not a join) since it's already readable
```

````python
# ── Define the query for vw_transfer_features ─────────────────────────────────

query_transfer = QueryDefinition(
    datasources=["transfer_value_observation"],
    columns=[
        # Identifier
        "transfer_value_observation.transfer_observation_id",
        # Human-readable names from lookup tables
        "player.player_name",
        "position.position_name",
        "nationality.nationality_name",
        "club.club_name",
        # Season (already readable as a year integer, no join needed)
        "transfer_value_observation.season_year",
        # Player features
        "transfer_value_observation.age_then_years",
        "transfer_value_observation.height_cm",
        # Performance stats
        "transfer_value_observation.total_games",
        "transfer_value_observation.total_goals",
        "transfer_value_observation.assists",
        "transfer_value_observation.penalty_kicks",
        "transfer_value_observation.total_minutes",
        # Club context (nullable — only present in 2019 sheet)
        "transfer_value_observation.club_performance",
        "transfer_value_observation.relegation",
        "transfer_value_observation.success_or_not",
        # Transfer value columns
        "transfer_value_observation.value_start_mln",
        "transfer_value_observation.value_delta_mln",
        # TARGET: end-of-season transfer value the ML model is trained to predict
        "transfer_value_observation.value_end_mln",
    ],
    joins=[
        # Join player lookup table
        JoinDefinition(
            type=JoinType.INNER,
            datasource="player",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.player_id",
                    foreign_column="player.player_id",
                )
            ],
        ),
        # Join position lookup table
        JoinDefinition(
            type=JoinType.INNER,
            datasource="position",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.position_id",
                    foreign_column="position.position_id",
                )
            ],
        ),
        # Join nationality lookup table
        JoinDefinition(
            type=JoinType.INNER,
            datasource="nationality",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.nationality_id",
                    foreign_column="nationality.nationality_id",
                )
            ],
        ),
        # Join club lookup table
        JoinDefinition(
            type=JoinType.INNER,
            datasource="club",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.club_id",
                    foreign_column="club.club_id",
                )
            ],
        ),
    ],
)

# ── Create the view (skips if it already exists) ───────────────────────────────
view_transfer = create_view_safe(
    client=client,
    database_id=DATABASE_ID,
    name=VIEW_TRANSFER,
    query=query_transfer,
)
````

## 8. View 3 — `vw_combined_player_value`

**Purpose:** Provides a unified schema across both datasets with a common set of columns
(player name, club, age, goals, assists, minutes played, and market value). Useful for
exploratory analysis and comparing patterns across both source datasets.
A `source` column identifies which dataset each row came from.

**Note:** DBRepo's `QueryDefinition` does not support UNION queries directly.
This view is therefore defined as a query over `forward_player_valuation` only,
with the shared columns. The equivalent for Dataset 2 is covered by `vw_transfer_features`.
A true UNION view is documented in the SQL file in the repository for reference,
but cannot be registered in DBRepo due to API limitations.

**Equivalent SQL (for reference only — stored in `/data/views.sql`):**
```sql
-- This UNION SQL is saved in the repo as documentation but cannot be
-- registered as a DBRepo view due to API limitations on UNION queries.
SELECT p.player_name, c.club_name,
       f.player_age_years AS age_years,
       f.goals, f.assists, f.minutes_played,
       f.market_value_mln AS market_value_mln,
       'forward_valuation' AS source
FROM forward_player_valuation f
INNER JOIN player p ON f.player_id = p.player_id
INNER JOIN club   c ON f.club_id   = c.club_id
UNION ALL
SELECT p.player_name, c.club_name,
       t.age_then_years AS age_years,
       t.total_goals AS goals, t.assists, t.total_minutes AS minutes_played,
       t.value_end_mln AS market_value_mln,
       'transfer_observation' AS source
FROM transfer_value_observation t
INNER JOIN player p ON t.player_id = p.player_id
INNER JOIN club   c ON t.club_id   = c.club_id;
```

````python
# ── Define the query for vw_combined_player_value ─────────────────────────────
#
# DBRepo does not support UNION in QueryDefinition.
# This view covers the shared columns from Dataset 1 (forward_player_valuation).
# The UNION SQL is stored in /data/views.sql in the repository for documentation.

query_combined = QueryDefinition(
    datasources=["forward_player_valuation"],
    columns=[
        # Common identifier fields across both datasets
        "player.player_name",
        "club.club_name",
        # Shared feature columns (same concept, mapped to Dataset 1 column names)
        "forward_player_valuation.player_age_years",   # age
        "forward_player_valuation.goals",
        "forward_player_valuation.assists",
        "forward_player_valuation.minutes_played",
        # Target variable (market value)
        "forward_player_valuation.market_value_mln",
    ],
    joins=[
        # Join player lookup
        JoinDefinition(
            type=JoinType.INNER,
            datasource="player",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.player_id",
                    foreign_column="player.player_id",
                )
            ],
        ),
        # Join club lookup
        JoinDefinition(
            type=JoinType.INNER,
            datasource="club",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.club_id",
                    foreign_column="club.club_id",
                )
            ],
        ),
    ],
)

# ── Create the view (skips if it already exists) ───────────────────────────────
view_combined = create_view_safe(
    client=client,
    database_id=DATABASE_ID,
    name=VIEW_COMBINED,
    query=query_combined,
)
````
