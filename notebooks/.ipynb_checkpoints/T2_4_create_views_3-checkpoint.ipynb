---
### How the DBRepo mapper works (and why both previous errors happened)

DBRepo's internal `query_to_subset` mapper has two strict rules:

1. **Every column must be in `table.column` format** — plain column names without the prefix raise `MalformedError`.
2. **The column list must include columns from ALL tables involved** (primary + all joined tables) — the mapper calls `numpy.array()` across each table's column list. If those lists have different lengths, numpy raises `ValueError: inhomogeneous shape`.

The fix: pass `table.column` for every column you want, from every table in the query (primary AND joined). This notebook fetches column names directly from DBRepo so nothing is hardcoded.

## 1. Imports

````python
from dbrepo.RestClient import RestClient
from dbrepo.api.dto import (
    QueryDefinition,
    JoinDefinition,
    JoinType,
    ConditionalDefinition,
)
from getpass import getpass
````

## 2. Configuration

````python
DBREPO_ENDPOINT = "https://test.dbrepo.tuwien.ac.at"
USERNAME        = "ekene"
DATABASE_ID     = "be8863ae-9794-4ee2-a964-3e22133b3840"

# View names
VIEW_FORWARD  = "vw_forward_features"
VIEW_TRANSFER = "vw_transfer_features"
VIEW_COMBINED = "vw_combined_player_value"

# Columns to EXCLUDE from views — these are internal FK/admin columns
# that are resolved via joins and should not appear as raw IDs in the output
EXCLUDE_FORWARD = {
    "forward_player_valuation": [
        "player_id",            # replaced by player.player_name via JOIN
        "club_id",              # replaced by club.club_name via JOIN
        "source_dataset_id",    # internal provenance — not an ML feature
        "original_player_name", # duplicate of player.player_name
        "original_team_name",   # duplicate of club.club_name
    ],
    "player": ["player_id"],    # PK not needed in output
    "club":   ["club_id"],      # PK not needed in output
}

EXCLUDE_TRANSFER = {
    "transfer_value_observation": [
        "player_id", "position_id", "nationality_id", "club_id",
        "source_dataset_id",
        "original_player_name", "original_position_name",
        "original_nationality_name", "original_club_name",
        # Keep only the _mln value columns for consistency
        "start_value_eur", "end_value_eur", "delta_value_eur",
        "age_now_years",    # less useful than age_then for ML
    ],
    "player":      ["player_id"],
    "position":    ["position_id"],
    "nationality": ["nationality_id"],
    "club":        ["club_id"],
}
````

## 3. Connect to DBRepo

````python
password = getpass(f"Enter DBRepo password for '{USERNAME}': ")

client = RestClient(
    endpoint=DBREPO_ENDPOINT,
    username=USERNAME,
    password=password,
)

print(f"Connected as: {client.whoami()}")
````

## 4. Fetch All Table Metadata from DBRepo
We fetch every table and build a lookup of `table_name → {column_name → column_object}`.
This is the only reliable way to get the exact column names and IDs DBRepo assigned.

````python
all_tables  = client.get_tables(database_id=DATABASE_ID)
table_by_name = {t.name: t for t in all_tables}

print("Tables in database:")
for name, tbl in table_by_name.items():
    col_names = [c.internal_name for c in tbl.columns]
    print(f"  {name} ({len(col_names)} cols): {col_names}")
````

## 5. Helper Functions

````python
def build_column_list(table_by_name, table_exclude_map):
    """
    Builds a list of 'table.column' strings for a QueryDefinition.

    Args:
        table_by_name:    dict of {table_name: table_object} from DBRepo
        table_exclude_map: dict of {table_name: [columns_to_exclude]}
                           Keys are the tables involved in this view (primary + joined).
                           Values are column names to omit from the view output.

    Returns:
        List of 'table.column' strings covering ALL involved tables.

    Why ALL tables?
        DBRepo's mapper internally calls numpy.array() on each table's column list.
        If the tables have different column counts the array is inhomogeneous and
        raises ValueError. Including columns from all joined tables in the column list
        ensures the mapper can resolve them uniformly.
    """
    columns = []
    for table_name, exclude_list in table_exclude_map.items():
        tbl = table_by_name[table_name]
        for col in tbl.columns:
            if col.internal_name not in exclude_list:
                columns.append(f"{table_name}.{col.internal_name}")
    return columns


def create_view_safe(client, database_id, name, query):
    """
    Creates a view only if it does not already exist.
    Skips silently if already present (safe to re-run).
    """
    existing = {v.name: v.id for v in client.get_views(database_id=database_id)}
    if name in existing:
        print(f"[SKIP] '{name}' already exists (id: {existing[name]})")
        return existing[name]

    view = client.create_view(
        database_id=database_id,
        name=name,
        query=query,
        is_public=True,
        is_schema_public=True,
    )
    print(f"[OK]   '{name}' created successfully (id: {view.id})")
    return view
````

## 6. View 1 — `vw_forward_features`

**Purpose:** Joins `forward_player_valuation` with `player` and `club` lookup tables.
One flat row per player with all features and the ML target variable.

**Source:** Dataset 1 — *Forward football player valuation* (Briseño & Rivera, 2024)  
**Target variable:** `market_value_mln`  
**Used in:** T2.6 ML pipeline data loading

**Equivalent SQL:**
```sql
SELECT
    f.forward_valuation_id, f.player_age_years, f.plays_in_europe,
    f.matches_played, f.goals, f.assists, f.minutes_per_goal,
    f.minutes_played, f.instagram_followers_mln, f.value_rank,
    f.market_value_mln,   -- TARGET
    p.player_name,
    c.club_name
FROM forward_player_valuation f
INNER JOIN player p ON f.player_id = p.player_id
INNER JOIN club   c ON f.club_id   = c.club_id;
```

````python
# Build the column list for ALL three tables involved in this view
# (primary: forward_player_valuation, joined: player, club)
forward_cols = build_column_list(table_by_name, EXCLUDE_FORWARD)
print("Columns for vw_forward_features:")
for c in forward_cols:
    print(f"  {c}")

query_forward = QueryDefinition(
    datasources=["forward_player_valuation"],
    columns=forward_cols,           # 'table.column' format, all tables included
    joins=[
        JoinDefinition(
            type=JoinType.INNER,
            datasource="player",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.player_id",
                    foreign_column="player.player_id",
                )
            ],
        ),
        JoinDefinition(
            type=JoinType.INNER,
            datasource="club",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.club_id",
                    foreign_column="club.club_id",
                )
            ],
        ),
    ],
)

view_forward = create_view_safe(
    client=client,
    database_id=DATABASE_ID,
    name=VIEW_FORWARD,
    query=query_forward,
)
````

## 7. View 2 — `vw_transfer_features`

**Purpose:** Joins `transfer_value_observation` with `player`, `position`, `nationality`,
and `club` lookup tables. One flat row per observation across all seasons.

**Source:** Dataset 2 — *Transfer Value Determinants* (Nisanov, 2025)  
**Target variable:** `value_end_mln`  
**Used in:** T2.6 ML pipeline data loading

**Equivalent SQL:**
```sql
SELECT
    t.transfer_observation_id, t.season_year, t.age_then_years, t.height_cm,
    t.total_games, t.total_goals, t.assists, t.penalty_kicks, t.total_minutes,
    t.club_performance, t.relegation, t.success_or_not,
    t.value_start_mln, t.value_delta_mln, t.value_end_mln,  -- TARGET
    p.player_name, pos.position_name, n.nationality_name, c.club_name
FROM transfer_value_observation t
INNER JOIN player      p   ON t.player_id      = p.player_id
INNER JOIN position    pos ON t.position_id    = pos.position_id
INNER JOIN nationality n   ON t.nationality_id = n.nationality_id
INNER JOIN club        c   ON t.club_id        = c.club_id;
```

````python
# Build the column list for ALL five tables involved in this view
transfer_cols = build_column_list(table_by_name, EXCLUDE_TRANSFER)
print("Columns for vw_transfer_features:")
for c in transfer_cols:
    print(f"  {c}")

query_transfer = QueryDefinition(
    datasources=["transfer_value_observation"],
    columns=transfer_cols,
    joins=[
        JoinDefinition(
            type=JoinType.INNER,
            datasource="player",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.player_id",
                    foreign_column="player.player_id",
                )
            ],
        ),
        JoinDefinition(
            type=JoinType.INNER,
            datasource="position",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.position_id",
                    foreign_column="position.position_id",
                )
            ],
        ),
        JoinDefinition(
            type=JoinType.INNER,
            datasource="nationality",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.nationality_id",
                    foreign_column="nationality.nationality_id",
                )
            ],
        ),
        JoinDefinition(
            type=JoinType.INNER,
            datasource="club",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.club_id",
                    foreign_column="club.club_id",
                )
            ],
        ),
    ],
)

view_transfer = create_view_safe(
    client=client,
    database_id=DATABASE_ID,
    name=VIEW_TRANSFER,
    query=query_transfer,
)
````

## 8. View 3 — `vw_combined_player_value`

**Purpose:** Shared-schema view over Dataset 1, exposing the columns common to both
datasets (player, club, age, goals, assists, minutes, market value).
Useful for exploratory analysis and comparing patterns across sources.

> **Note:** DBRepo does not support `UNION ALL` in `QueryDefinition`.
> The full UNION SQL is saved in `/data/views.sql` in the repository for reference.

**Equivalent SQL (reference — see `/data/views.sql`):**
```sql
SELECT p.player_name, c.club_name, f.player_age_years AS age_years,
       f.goals, f.assists, f.minutes_played, f.market_value_mln, 'dataset_1' AS source
FROM forward_player_valuation f
INNER JOIN player p ON f.player_id = p.player_id
INNER JOIN club   c ON f.club_id   = c.club_id
UNION ALL
SELECT p.player_name, c.club_name, t.age_then_years AS age_years,
       t.total_goals AS goals, t.assists, t.total_minutes AS minutes_played,
       t.value_end_mln, 'dataset_2' AS source
FROM transfer_value_observation t
INNER JOIN player p ON t.player_id = p.player_id
INNER JOIN club   c ON t.club_id   = c.club_id;
```

````python
# Shared columns that exist in both datasets, using Dataset 1 column names.
# Only the intersection of meaningful features is included.
EXCLUDE_COMBINED = {
    "forward_player_valuation": [
        # Exclude everything except the shared feature columns
        "forward_valuation_id",
        "source_dataset_id",
        "player_id",
        "club_id",
        "original_player_name",
        "original_team_name",
        "plays_in_europe",          # no equivalent in Dataset 2
        "matches_played",           # Dataset 2 uses total_games
        "minutes_per_goal",         # derived stat, no equivalent in Dataset 2
        "instagram_followers_mln",  # no equivalent in Dataset 2
        "value_rank",               # no equivalent in Dataset 2
        # market_value_mln is KEPT as the target
    ],
    "player": ["player_id"],
    "club":   ["club_id"],
}

combined_cols = build_column_list(table_by_name, EXCLUDE_COMBINED)
print("Columns for vw_combined_player_value:")
for c in combined_cols:
    print(f"  {c}")

query_combined = QueryDefinition(
    datasources=["forward_player_valuation"],
    columns=combined_cols,
    joins=[
        JoinDefinition(
            type=JoinType.INNER,
            datasource="player",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.player_id",
                    foreign_column="player.player_id",
                )
            ],
        ),
        JoinDefinition(
            type=JoinType.INNER,
            datasource="club",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.club_id",
                    foreign_column="club.club_id",
                )
            ],
        ),
    ],
)

view_combined = create_view_safe(
    client=client,
    database_id=DATABASE_ID,
    name=VIEW_COMBINED,
    query=query_combined,
)
````

## 9. Verify — List All Views

````python
all_views   = client.get_views(database_id=DATABASE_ID)
view_id_map = {v.name: v.id for v in all_views}

print(f"Views in database '{DATABASE_ID}':")
print("-" * 60)
for v in all_views:
    print(f"  Name : {v.name}")
    print(f"  ID   : {v.id}")
    print()

missing = {VIEW_FORWARD, VIEW_TRANSFER, VIEW_COMBINED} - set(view_id_map)
if missing:
    print(f"WARNING: views not found: {missing}")
else:
    print("All three views registered successfully.")
````
