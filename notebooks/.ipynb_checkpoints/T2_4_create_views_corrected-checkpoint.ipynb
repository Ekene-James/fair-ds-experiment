```
from dbrepo.RestClient import RestClient
from dbrepo.enums import JoinType
from dbrepo.model import (
    QueryDefinition,
    JoinDefinition,
    ConditionalDefinition,
)

import pandas as pd
from getpass import getpass
```

```
BASE_URL = "https://test.dbrepo.tuwien.ac.at"

USERNAME = "ekene"
DATABASE_ID = "be8863ae-9794-4ee2-a964-3e22133b3840"

PASSWORD = getpass("DBRepo Password: ")
```

```
client = RestClient(BASE_URL)
client.login(USERNAME, PASSWORD)

print("Connected to DBRepo")
```

```
def get_tables_by_name(client, database_id):

    tables = client.get_tables(database_id)

    table_by_name = {}

    for t in tables:
        full_table = client.get_table(
            database_id,
            t.id,
        )

        # IMPORTANT FIX:
        # use internal_name consistently
        table_by_name[full_table.internal_name] = full_table

    return table_by_name
```

```
def create_view_safe(client, database_id, name, query):

    existing_views = client.get_views(database_id)

    existing = {
        v.name: v.id
        for v in existing_views
    }

    if name in existing:
        print(f"[SKIP] '{name}' already exists (id: {existing[name]})")
        return existing[name]

    view = client.create_view(
        database_id=database_id,
        name=name,
        query=query,
        is_public=True,
        is_schema_public=True,
    )

    print(f"[OK] '{name}' created (id: {view.id})")

    return view.id
```

```
def get_primary_columns(
    table_by_name,
    table_name,
    exclude=None,
):

    if exclude is None:
        exclude = []

    tbl = table_by_name[table_name]

    return [
        f"{table_name}.{col.internal_name}"
        for col in tbl.columns
        if col.internal_name not in exclude
    ]
```

```
table_by_name = get_tables_by_name(
    client,
    DATABASE_ID,
)

print("Available tables:")

for name in table_by_name:
    print("-", name)
```

## View 1 — Forward Features

```
VIEW_FORWARD = "vw_forward_features"

forward_primary_cols = get_primary_columns(
    table_by_name,
    "forward_player_valuation",
    exclude=[
        "player_id",
        "club_id",
        "source_dataset_id",
        "original_player_name",
        "original_team_name",
    ],
)

# IMPORTANT:
# DBRepo expects a FLAT LIST of fully-qualified strings.
# NOT nested lists.

extra_forward_cols = [
    "player.player_name",
    "player.birth_date",
    "club.club_name",
    "source_dataset.dataset_name",
]

forward_cols = (
    forward_primary_cols
    + extra_forward_cols
)

# Validation
assert all(isinstance(c, str) for c in forward_cols)

query_forward = QueryDefinition(
    datasources=[
        "forward_player_valuation"
    ],

    columns=forward_cols,

    joins=[
        JoinDefinition(
            type=JoinType.INNER,
            datasource="player",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.player_id",
                    foreign_column="player.player_id",
                )
            ],
        ),

        JoinDefinition(
            type=JoinType.LEFT,
            datasource="club",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.club_id",
                    foreign_column="club.club_id",
                )
            ],
        ),

        JoinDefinition(
            type=JoinType.LEFT,
            datasource="source_dataset",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.source_dataset_id",
                    foreign_column="source_dataset.source_dataset_id",
                )
            ],
        ),
    ],
)

create_view_safe(
    client,
    DATABASE_ID,
    VIEW_FORWARD,
    query_forward,
)
```

## View 2 — Transfer Features

```
VIEW_TRANSFER = "vw_transfer_features"

transfer_primary_cols = get_primary_columns(
    table_by_name,
    "transfer_value_observation",
    exclude=[
        "player_id",
        "club_id",
        "position_id",
        "nationality_id",
        "source_dataset_id",
    ],
)

extra_transfer_cols = [
    "player.player_name",
    "club.club_name",
    "position.position_name",
    "nationality.country_name",
    "source_dataset.dataset_name",
]

transfer_cols = (
    transfer_primary_cols
    + extra_transfer_cols
)

assert all(isinstance(c, str) for c in transfer_cols)

query_transfer = QueryDefinition(
    datasources=[
        "transfer_value_observation"
    ],

    columns=transfer_cols,

    joins=[
        JoinDefinition(
            type=JoinType.INNER,
            datasource="player",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.player_id",
                    foreign_column="player.player_id",
                )
            ],
        ),

        JoinDefinition(
            type=JoinType.LEFT,
            datasource="club",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.club_id",
                    foreign_column="club.club_id",
                )
            ],
        ),

        JoinDefinition(
            type=JoinType.LEFT,
            datasource="position",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.position_id",
                    foreign_column="position.position_id",
                )
            ],
        ),

        JoinDefinition(
            type=JoinType.LEFT,
            datasource="nationality",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.nationality_id",
                    foreign_column="nationality.nationality_id",
                )
            ],
        ),

        JoinDefinition(
            type=JoinType.LEFT,
            datasource="source_dataset",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.source_dataset_id",
                    foreign_column="source_dataset.source_dataset_id",
                )
            ],
        ),
    ],
)

create_view_safe(
    client,
    DATABASE_ID,
    VIEW_TRANSFER,
    query_transfer,
)
```

```
views = client.get_views(DATABASE_ID)

print("Created views:")

for v in views:
    print(v.name)
```
