## Purpose
This notebook creates views in DBRepo that de-normalise the 3NF schema (Student A, T2.1)
into flat, query-ready tables for the ML pipeline.

Three views are created:
1. **`vw_forward_features`** – Flat table from Dataset 1, predicting `market_value_mln`
2. **`vw_transfer_features`** – Flat table from Dataset 2, predicting `value_end_mln`
3. **`vw_combined_player_value`** – Shared schema from Dataset 1 for cross-dataset analysis

---
### Fix note
DBRepo's `QueryDefinition` mapper calls `numpy.array()` on each table's column list internally.
When multiple joined tables have **different numbers of columns**, NumPy raises a `ValueError`
because the resulting array is inhomogeneous (jagged). The fix is:
- Pass column names **without** the `table.` prefix
- Only list columns from the **primary (FROM) table** — joined table columns are resolved by DBRepo automatically via the join conditions

## 4. Look Up Table IDs Automatically
DBRepo assigns a UUID to each table at creation time. We fetch them by name so nothing is hardcoded.

````python
all_tables = client.get_tables(database_id=DATABASE_ID)
table_ids  = {t.name: t.id for t in all_tables}

print("Tables found in database:")
for name, tid in table_ids.items():
    print(f"  {name}: {tid}")

# Fail early if any expected table is missing
missing = [t for t in TABLE_NAMES if t not in table_ids]
if missing:
    raise RuntimeError(
        f"Missing tables: {missing}\n"
        "Ensure Student A has created all tables (T2.1) before running this notebook."
    )

print("\nAll expected tables present. Ready to create views.")
````

## 5. Helper — Look Up Column Names from DBRepo

Because we must pass **only column names without the table prefix**, and only from
the **primary table**, this helper fetches the actual column names DBRepo knows about
for a given table. This avoids any mismatch with the schema as registered.

````python
def get_column_names(client, database_id, table_name, table_ids, exclude=None):
    """
    Fetch the column names for a table from DBRepo.
    
    Args:
        exclude: list of column names to omit (e.g. FK id columns not needed in view)
    Returns:
        list of column internal_name strings
    """
    exclude = exclude or []
    table   = client.get_table(database_id=database_id, table_id=table_ids[table_name])
    return [col.internal_name for col in table.columns if col.internal_name not in exclude]


def create_view_safe(client, database_id, name, query):
    """
    Creates a view only if it does not already exist.
    Prints status and returns the view object either way.
    """
    existing = {v.name: v.id for v in client.get_views(database_id=database_id)}

    if name in existing:
        print(f"[SKIP] '{name}' already exists (id: {existing[name]})")
        return existing[name]

    view = client.create_view(
        database_id=database_id,
        name=name,
        query=query,
        is_public=True,
        is_schema_public=True,
    )
    print(f"[OK]   '{name}' created (id: {view.id})")
    return view


# Preview columns available in each primary table
print("Columns in forward_player_valuation:")
print(get_column_names(client, DATABASE_ID, "forward_player_valuation", table_ids))
print()
print("Columns in transfer_value_observation:")
print(get_column_names(client, DATABASE_ID, "transfer_value_observation", table_ids))
````

## 6. View 1 — `vw_forward_features`

**Purpose:** Joins `forward_player_valuation` with `player` and `club` lookup tables.
One flat row per player with all features and the ML target variable.

**Source:** Dataset 1 — *Forward football player valuation* (Briseño & Rivera, 2024)  
**Target variable:** `market_value_mln`  
**Used in:** T2.6 ML pipeline data loading

**Equivalent SQL:**
```sql
SELECT
    f.forward_valuation_id, f.player_age_years, f.plays_in_europe,
    f.matches_played, f.goals, f.assists, f.minutes_per_goal,
    f.minutes_played, f.instagram_followers_mln, f.value_rank,
    f.market_value_mln,   -- TARGET
    p.player_name,
    c.club_name
FROM forward_player_valuation f
INNER JOIN player p ON f.player_id = p.player_id
INNER JOIN club   c ON f.club_id   = c.club_id;
```

> **Why no column prefix / joined columns?**  
> DBRepo's internal mapper uses `numpy.array()` on each table's column list. When joined
> tables have different column counts the array becomes inhomogeneous and raises a `ValueError`.
> Passing only the primary table's column names (no `table.` prefix) avoids this bug.
> DBRepo automatically includes the join columns in the resulting view.

````python
# Fetch primary table columns from DBRepo, excluding internal FK id columns
# (player_id, club_id, source_dataset_id) since those are resolved via the joins
forward_cols = get_column_names(
    client, DATABASE_ID, "forward_player_valuation", table_ids,
    exclude=["player_id", "club_id", "source_dataset_id",
             "original_player_name", "original_team_name"],
)
print("Columns selected for vw_forward_features:", forward_cols)

query_forward = QueryDefinition(
    datasources=["forward_player_valuation"],
    # Only primary table column names, no 'table.' prefix
    columns=forward_cols,
    joins=[
        # Bring in player_name from the player lookup table
        JoinDefinition(
            type=JoinType.INNER,
            datasource="player",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.player_id",
                    foreign_column="player.player_id",
                )
            ],
        ),
        # Bring in club_name from the club lookup table
        JoinDefinition(
            type=JoinType.INNER,
            datasource="club",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.club_id",
                    foreign_column="club.club_id",
                )
            ],
        ),
    ],
)

view_forward = create_view_safe(
    client=client,
    database_id=DATABASE_ID,
    name=VIEW_FORWARD,
    query=query_forward,
)
````

## 7. View 2 — `vw_transfer_features`

**Purpose:** Joins `transfer_value_observation` with `player`, `position`, `nationality`,
and `club` lookup tables. One flat row per observation across all seasons.

**Source:** Dataset 2 — *Transfer Value Determinants* (Nisanov, 2025)  
**Target variable:** `value_end_mln`  
**Used in:** T2.6 ML pipeline data loading

**Equivalent SQL:**
```sql
SELECT
    t.transfer_observation_id, t.season_year, t.age_then_years, t.height_cm,
    t.total_games, t.total_goals, t.assists, t.penalty_kicks, t.total_minutes,
    t.club_performance, t.relegation, t.success_or_not,
    t.value_start_mln, t.value_delta_mln,
    t.value_end_mln,   -- TARGET
    p.player_name, pos.position_name, n.nationality_name, c.club_name
FROM transfer_value_observation t
INNER JOIN player      p   ON t.player_id      = p.player_id
INNER JOIN position    pos ON t.position_id    = pos.position_id
INNER JOIN nationality n   ON t.nationality_id = n.nationality_id
INNER JOIN club        c   ON t.club_id        = c.club_id;
```

````python
# Fetch primary table columns, excluding FK id columns resolved via joins
transfer_cols = get_column_names(
    client, DATABASE_ID, "transfer_value_observation", table_ids,
    exclude=["player_id", "position_id", "nationality_id", "club_id",
             "source_dataset_id", "original_player_name", "original_position_name",
             "original_nationality_name", "original_club_name",
             # Exclude raw EUR columns — use the _mln versions for consistency
             "start_value_eur", "end_value_eur", "delta_value_eur",
             "age_now_years"],  # age_now is less useful for ML than age_then
)
print("Columns selected for vw_transfer_features:", transfer_cols)

query_transfer = QueryDefinition(
    datasources=["transfer_value_observation"],
    columns=transfer_cols,
    joins=[
        # player_name
        JoinDefinition(
            type=JoinType.INNER,
            datasource="player",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.player_id",
                    foreign_column="player.player_id",
                )
            ],
        ),
        # position_name
        JoinDefinition(
            type=JoinType.INNER,
            datasource="position",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.position_id",
                    foreign_column="position.position_id",
                )
            ],
        ),
        # nationality_name
        JoinDefinition(
            type=JoinType.INNER,
            datasource="nationality",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.nationality_id",
                    foreign_column="nationality.nationality_id",
                )
            ],
        ),
        # club_name
        JoinDefinition(
            type=JoinType.INNER,
            datasource="club",
            conditionals=[
                ConditionalDefinition(
                    column="transfer_value_observation.club_id",
                    foreign_column="club.club_id",
                )
            ],
        ),
    ],
)

view_transfer = create_view_safe(
    client=client,
    database_id=DATABASE_ID,
    name=VIEW_TRANSFER,
    query=query_transfer,
)
````

## 8. View 3 — `vw_combined_player_value`

**Purpose:** A shared-schema view over Dataset 1 exposing the columns common to both
datasets (player, club, age, goals, assists, minutes, market value). Useful for
exploratory analysis across both sources.

**Limitation:** DBRepo's `QueryDefinition` does not support `UNION ALL`.
The full UNION SQL is saved in `/data/views.sql` in the repository for reference
and documentation, but cannot be registered as a DBRepo view.

**Equivalent SQL (reference only — see `/data/views.sql`):**
```sql
SELECT p.player_name, c.club_name,
       f.player_age_years AS age_years, f.goals, f.assists,
       f.minutes_played,  f.market_value_mln, 'dataset_1' AS source
FROM forward_player_valuation f
INNER JOIN player p ON f.player_id = p.player_id
INNER JOIN club   c ON f.club_id   = c.club_id
UNION ALL
SELECT p.player_name, c.club_name,
       t.age_then_years AS age_years, t.total_goals AS goals, t.assists,
       t.total_minutes AS minutes_played, t.value_end_mln, 'dataset_2' AS source
FROM transfer_value_observation t
INNER JOIN player p ON t.player_id = p.player_id
INNER JOIN club   c ON t.club_id   = c.club_id;
```

````python
# Shared columns that exist in both datasets (using Dataset 1 column names)
# This is the intersection of meaningful features across both sources
combined_cols = [
    "player_age_years",       # = age_then_years in Dataset 2
    "goals",                  # = total_goals in Dataset 2
    "assists",                # same name in both
    "minutes_played",         # = total_minutes in Dataset 2
    "market_value_mln",       # = value_end_mln in Dataset 2  (TARGET)
]
print("Columns for vw_combined_player_value:", combined_cols)

query_combined = QueryDefinition(
    datasources=["forward_player_valuation"],
    columns=combined_cols,
    joins=[
        JoinDefinition(
            type=JoinType.INNER,
            datasource="player",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.player_id",
                    foreign_column="player.player_id",
                )
            ],
        ),
        JoinDefinition(
            type=JoinType.INNER,
            datasource="club",
            conditionals=[
                ConditionalDefinition(
                    column="forward_player_valuation.club_id",
                    foreign_column="club.club_id",
                )
            ],
        ),
    ],
)

view_combined = create_view_safe(
    client=client,
    database_id=DATABASE_ID,
    name=VIEW_COMBINED,
    query=query_combined,
)
````
