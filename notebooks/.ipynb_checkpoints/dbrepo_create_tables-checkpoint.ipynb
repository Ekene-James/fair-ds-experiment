## Purpose
Creates all 8 tables in DBRepo matching the 3NF schema defined in `schema.sql`.
Uses the DBRepo Python SDK `create_table()` which infers column types from
a pandas DataFrame's dtypes — so dtype choices here are critical.

## Pandas dtype → DBRepo/MariaDB type mapping
| Pandas dtype | DBRepo creates |
|---|---|
| `int64` | INT |
| `Int64` (capital I, nullable) | INT (nullable) |
| `float64` | DOUBLE/DECIMAL |
| `bool` | BOOLEAN |
| `string` | VARCHAR |

## Schema SQL vs previous notebook — what was wrong
The previous notebook used `float64` for `club_performance` and `success_or_not`,
which caused DBRepo to create them as DECIMAL instead of the correct SMALLINT (INT).
This notebook fixes all dtype mappings to match `schema.sql` exactly.

## Tables created (in FK dependency order)
```
source_dataset → player → club → position → nationality → season
    ↓                                                        ↓
forward_player_valuation              transfer_value_observation
```

## 4. Define Table Schemas

Each schema is defined as a **one-row seed DataFrame**.
DBRepo infers the column types from the DataFrame dtypes — the actual seed
values don't matter, only the dtypes do. The seed row is not committed
to the database (`with_data=False`).

### Critical dtype rules (learned from failed uploads):
- Use `pd.Int64Dtype()` (capital I) for **nullable integer** columns
  (`club_performance`, `success_or_not`) — these are SMALLINT NULL in the schema.
  Using `float64` here causes DBRepo to create them as DECIMAL, which then
  rejects integer CSV values on upload.
- Use `bool` for BOOLEAN columns (`plays_in_europe`, `relegation`).
- Use `float64` only for genuine DECIMAL columns
  (`market_value_mln`, `instagram_followers_mln`, `height_cm`, value columns).
- Use `pd.StringDtype()` for all VARCHAR/TEXT columns.

````python
# Shorthand helpers for cleaner schema definitions
INT   = pd.Series([0],    dtype='int64')          # INT / BIGINT / SMALLINT (not nullable)
INTN  = pd.Series([0],    dtype=pd.Int64Dtype())   # SMALLINT NULL / INT NULL (nullable integer)
FLT   = pd.Series([0.0],  dtype='float64')         # DECIMAL / DOUBLE
FLTN  = pd.Series([None], dtype='float64')         # DECIMAL NULL
BOOL  = pd.Series([True], dtype='bool')            # BOOLEAN
STR   = pd.Series(['x'],  dtype=pd.StringDtype())  # VARCHAR / TEXT

# ── 1. source_dataset ─────────────────────────────────────────────────────────
# Schema: SMALLINT PK, 8x VARCHAR NOT NULL, 1x VARCHAR nullable, 1x TEXT nullable
source_dataset_seed = pd.DataFrame({
    'source_dataset_id': INT,
    'dataset_title':     STR,
    'original_creators': STR,
    'publisher':         STR,
    'version_label':     STR,   # nullable in schema but VARCHAR — string handles both
    'doi':               STR,
    'doi_url':           STR,
    'license_name':      STR,
    'license_url':       STR,
    'source_filename':   STR,   # nullable in schema
    'notes':             STR,   # TEXT nullable in schema
})

# ── 2. player ─────────────────────────────────────────────────────────────────
# Schema: INT AUTO_INCREMENT PK, VARCHAR NOT NULL UNIQUE
player_seed = pd.DataFrame({
    'player_id':   INT,
    'player_name': STR,
})

# ── 3. club ───────────────────────────────────────────────────────────────────
club_seed = pd.DataFrame({
    'club_id':   INT,
    'club_name': STR,
})

# ── 4. position ───────────────────────────────────────────────────────────────
position_seed = pd.DataFrame({
    'position_id':   INT,
    'position_name': STR,
})

# ── 5. nationality ────────────────────────────────────────────────────────────
nationality_seed = pd.DataFrame({
    'nationality_id':   INT,
    'nationality_name': STR,
})

# ── 6. season ─────────────────────────────────────────────────────────────────
# Schema: SMALLINT PK, VARCHAR nullable
# Note: schema.sql pre-inserts 5 rows (2019-2023) — do NOT upload season CSV
season_seed = pd.DataFrame({
    'season_year': INT,
    'notes':       STR,
})

# ── 7. forward_player_valuation ───────────────────────────────────────────────
# Schema: INT PK, FK ids (INT), VARCHAR, SMALLINT, DECIMAL, BOOLEAN, INT cols
# plays_in_europe: BOOLEAN — use bool dtype
# instagram_followers_mln: DECIMAL(12,3) — use float64
forward_seed = pd.DataFrame({
    'forward_valuation_id':    INT,
    'source_dataset_id':       INT,
    'player_id':               INT,
    'club_id':                 INT,
    'original_player_name':    STR,
    'original_team_name':      STR,
    'player_age_years':        INT,   # SMALLINT
    'market_value_mln':        FLT,   # DECIMAL(12,3)
    'value_rank':              INT,   # SMALLINT
    'plays_in_europe':         BOOL,  # BOOLEAN
    'matches_played':          INT,
    'goals':                   INT,
    'assists':                 INT,
    'minutes_per_goal':        INT,
    'minutes_played':          INT,
    'instagram_followers_mln': FLT,   # DECIMAL(12,3)
})

# ── 8. transfer_value_observation ─────────────────────────────────────────────
# Schema: BIGINT PK, FK ids (INT), SMALLINT cols, BOOLEAN NULL, SMALLINT NULL,
#         INT cols, DECIMAL cols
#
# CRITICAL FIXES vs old notebook:
#   club_performance: SMALLINT NULL → use INTN (pd.Int64Dtype), NOT float64
#   success_or_not:   SMALLINT NULL → use INTN (pd.Int64Dtype), NOT float64
#   relegation:       BOOLEAN NULL  → use bool (DBRepo handles nullable BOOLEAN)
#
# Using float64 for these caused DBRepo to create them as DECIMAL,
# which then rejected integer values on CSV upload.
transfer_seed = pd.DataFrame({
    'transfer_observation_id':   INT,
    'source_dataset_id':         INT,
    'player_id':                 INT,
    'position_id':               INT,
    'nationality_id':            INT,
    'club_id':                   INT,
    'season_year':               INT,   # SMALLINT FK
    'original_player_name':      STR,
    'original_position_name':    STR,
    'original_nationality_name': STR,
    'original_club_name':        STR,
    'age_then_years':            INT,   # SMALLINT
    'age_now_years':             INT,   # SMALLINT
    'club_performance':          INTN,  # SMALLINT NULL ← was float64, now Int64
    'relegation':                BOOL,  # BOOLEAN NULL
    'success_or_not':            INTN,  # SMALLINT NULL ← was float64, now Int64
    'total_games':               INT,
    'assists':                   INT,
    'penalty_kicks':             INT,
    'total_minutes':             INT,
    'total_goals':               INT,
    'height_cm':                 FLT,   # DECIMAL(6,2)
    'start_value_eur':           FLTN,  # DECIMAL(16,2) NULL
    'end_value_eur':             FLTN,  # DECIMAL(16,2) NULL
    'delta_value_eur':           FLT,   # DECIMAL(16,2)
    'value_start_mln':           FLT,   # DECIMAL(12,3)
    'value_end_mln':             FLT,   # DECIMAL(12,3)
    'value_delta_mln':           FLT,   # DECIMAL(12,3)
})

print("Schema seed DataFrames defined.")
print("\nColumn counts:")
for name, df in [
    ('source_dataset', source_dataset_seed),
    ('player', player_seed),
    ('club', club_seed),
    ('position', position_seed),
    ('nationality', nationality_seed),
    ('season', season_seed),
    ('forward_player_valuation', forward_seed),
    ('transfer_value_observation', transfer_seed),
]:
    print(f"  {name}: {len(df.columns)} columns")
````

## 5. Table Descriptions
These appear as metadata in DBRepo making the database citable and searchable.

````python
table_descriptions = {
    'source_dataset': (
        'Provenance metadata for the two reused external source datasets. '
        'Records original creators, publisher (Mendeley Data), DOI, licence (CC BY 4.0), '
        'and source filename. The student group is not the original publisher or rights holder.'
    ),
    'player': (
        'Deduplicated lookup table of football player names appearing across both source datasets. '
        'Referenced by forward_player_valuation and transfer_value_observation via player_id.'
    ),
    'club': (
        'Deduplicated lookup table of football club names appearing across both source datasets. '
        'Referenced by forward_player_valuation and transfer_value_observation via club_id.'
    ),
    'position': (
        'Lookup table of football positions (e.g. Striker, Defensive Midfield) '
        'from the Transfer Value Determinants dataset. '
        'Referenced by transfer_value_observation via position_id.'
    ),
    'nationality': (
        'Lookup table of player nationality labels from the Transfer Value Determinants dataset. '
        'Referenced by transfer_value_observation via nationality_id.'
    ),
    'season': (
        'Season years (2019-2023) from the Transfer Value Determinants workbook sheets. '
        'Pre-populated by the schema INSERT statements. '
        'Referenced by transfer_value_observation via season_year.'
    ),
    'forward_player_valuation': (
        'Normalised observations from the Forward Football Player Valuation dataset '
        '(Briseño & Rivera, 2024, DOI: 10.17632/cgc33scxg7.1). '
        'One row per forward player. Contains age, match statistics (goals, assists, minutes), '
        'Instagram followers, and market value in millions EUR. '
        'Target variable for ML: market_value_mln.'
    ),
    'transfer_value_observation': (
        'Normalised observations from the Transfer Value Determinants dataset '
        '(Nisanov, 2025, DOI: 10.17632/3btg6ptc7b.2). '
        'One row per player-season observation across 2019-2023. '
        'Contains performance stats, height, club context, and start/end transfer values. '
        'Target variable for ML: value_end_mln. '
        'club_performance, relegation, success_or_not are NULL for seasons 2020-2023.'
    ),
}
````

## 6. Create Tables
Tables are created in FK dependency order — lookup tables before fact tables.
Each table is checked first so the notebook is safe to re-run.

````python
# Build a name → id map of tables that already exist
existing = {
    t.name: t.id
    for t in client.get_tables(database_id=DATABASE_ID)
}
print(f"Tables already in database: {list(existing.keys()) or 'none'}")

# Define creation order — must respect FK dependencies
creation_order = [
    ('source_dataset',             source_dataset_seed),
    ('player',                     player_seed),
    ('club',                       club_seed),
    ('position',                   position_seed),
    ('nationality',                nationality_seed),
    ('season',                     season_seed),
    ('forward_player_valuation',   forward_seed),
    ('transfer_value_observation', transfer_seed),
]

created_table_ids = dict(existing)  # start with already-existing ones

for table_name, seed_df in creation_order:
    if table_name in existing:
        print(f"[SKIP] '{table_name}' already exists (id: {existing[table_name]})")
        continue

    try:
        table = client.create_table(
            database_id=DATABASE_ID,
            name=table_name,
            dataframe=seed_df,
            description=table_descriptions[table_name],
            is_public=IS_PUBLIC,
            is_schema_public=IS_SCHEMA_PUBLIC,
            with_data=False,  # seed row defines schema only — no data committed
        )
        created_table_ids[table_name] = table.id
        print(f"[OK]   '{table_name}' created (id: {table.id})")

    except Exception as e:
        print(f"[FAIL] '{table_name}': {type(e).__name__}: {e}")

print()
print(f"Done. {len(created_table_ids)}/8 tables available.")
````

## 7. Verify — Check Column Types in DBRepo
Fetches each table from DBRepo and prints the column types it actually created.
Compare against the schema.sql to confirm everything matches.

````python
# Expected types from schema.sql for spot-checking the critical columns
expected_types = {
    'forward_player_valuation': {
        'plays_in_europe':         'BOOLEAN',
        'goals':                   'INT',
        'assists':                 'INT',
        'market_value_mln':        'DECIMAL',
        'instagram_followers_mln': 'DECIMAL',
    },
    'transfer_value_observation': {
        'club_performance': 'INT',      # SMALLINT maps to INT in DBRepo
        'relegation':       'BOOLEAN',
        'success_or_not':   'INT',      # SMALLINT maps to INT
        'height_cm':        'DECIMAL',
        'start_value_eur':  'DECIMAL',
    },
}

all_tables_brief = client.get_tables(database_id=DATABASE_ID)

print("Column types as registered in DBRepo:")
print("=" * 60)

for t_brief in all_tables_brief:
    full = client.get_table(database_id=DATABASE_ID, table_id=t_brief.id)
    print(f"\n{full.name}:")
    for col in full.columns:
        # Flag any columns that look wrong based on expected types
        flag = ""
        if full.name in expected_types:
            if col.internal_name in expected_types[full.name]:
                exp = expected_types[full.name][col.internal_name]
                if exp.upper() not in str(col.type).upper():
                    flag = f"  ← EXPECTED {exp}, got {col.type}"
        print(f"  {col.internal_name}: {col.type}{flag}")
````

## 9. Schema SQL Compatibility Notes

The `schema.sql` file is correct and does not need editing. Notes for reference:

**What the schema SQL does that DBRepo SDK cannot do:**
- `AUTO_INCREMENT` on primary keys — DBRepo handles this internally
- `UNIQUE KEY` constraints — not enforced via SDK `create_table`, enforced at DB level
- `FOREIGN KEY` constraints — not added via SDK, enforced at DB level
- `INSERT INTO source_dataset` — pre-loads provenance rows
- `INSERT INTO season` — pre-loads 2019-2023 rows (do NOT re-upload season CSV)

**Data fit check (all values within schema type ranges):**
- `market_value_mln`: 7.0–40.0 → fits DECIMAL(12,3) ✓
- `instagram_followers_mln`: 0.0–635.0 → fits DECIMAL(12,3) ✓
- `club_performance`: -10 to 13 → fits SMALLINT (-32768 to 32767) ✓
- `success_or_not`: -1, 0, 1 → fits SMALLINT ✓
- `height_cm`: integer values → fits DECIMAL(6,2) ✓
- `start_value_eur`: up to 25,000,000 → fits DECIMAL(16,2) ✓
