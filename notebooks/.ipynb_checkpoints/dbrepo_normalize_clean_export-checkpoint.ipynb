
# DBRepo Normalization + Clean CSV Export Pipeline


```

import pandas as pd
from pathlib import Path
```

```

DATA_DIR = Path("../data")

FORWARD_FILE = DATA_DIR / "Dataset soccerplayers(1).xlsx"
TRANSFER_FILE = DATA_DIR / "Nisanov_final_data(1).xlsx"

EXPORT_DIR = Path("../normalized_exports")
EXPORT_DIR.mkdir(exist_ok=True)
```

```

forward_df = pd.read_excel(FORWARD_FILE)

transfer_sheets = pd.read_excel(
    TRANSFER_FILE,
    sheet_name=None,
)

transfer_df = pd.concat(
    [
        sheet.dropna(axis=1, how="all")
        for sheet in transfer_sheets.values()
    ],
    ignore_index=True,
    sort=False,
)

forward_df.columns = [
    c.strip().lower()
    for c in forward_df.columns
]

transfer_df.columns = [
    c.strip().lower()
    for c in transfer_df.columns
]

print(forward_df.shape)
print(transfer_df.shape)
```

```

encoding_fixes = {
    "脙聡aglar Soyuncu": "Çağlar Söyüncü",
}

forward_df = forward_df.replace(encoding_fixes)
transfer_df = transfer_df.replace(encoding_fixes)
```

```

player_names = pd.concat([
    forward_df["nombre"].rename("player_name"),
    transfer_df["player_name"],
])

player_df = (
    player_names
    .dropna()
    .drop_duplicates()
    .reset_index(drop=True)
    .to_frame()
)

player_df["player_id"] = range(1, len(player_df) + 1)

player_df = player_df[
    ["player_id", "player_name"]
]
```

```

club_names = pd.concat([
    forward_df["team"].rename("club_name"),
    transfer_df["club_name"],
])

club_df = (
    club_names
    .dropna()
    .drop_duplicates()
    .reset_index(drop=True)
    .to_frame()
)

club_df["club_id"] = range(1, len(club_df) + 1)

club_df = club_df[
    ["club_id", "club_name"]
]
```

```

position_df = (
    transfer_df[["position"]]
    .dropna()
    .drop_duplicates()
    .reset_index(drop=True)
)

position_df["position_id"] = range(1, len(position_df) + 1)

position_df = position_df[
    ["position_id", "position"]
]
```

```

nationality_df = (
    transfer_df[["nationality"]]
    .dropna()
    .drop_duplicates()
    .reset_index(drop=True)
)

nationality_df["nationality_id"] = range(1, len(nationality_df) + 1)

nationality_df = nationality_df[
    ["nationality_id", "nationality"]
]
```

```

source_dataset_df = pd.DataFrame({
    "source_dataset_id": [1, 2],
    "dataset_name": [
        "Forward football player valuation",
        "Transfer Value Determinants",
    ],
})
```

```

forward_norm = forward_df.copy()

forward_norm = forward_norm.merge(
    player_df,
    left_on="nombre",
    right_on="player_name",
    how="left",
)

forward_norm = forward_norm.merge(
    club_df,
    left_on="team",
    right_on="club_name",
    how="left",
)

forward_fact = pd.DataFrame({
    "forward_valuation_id": range(1, len(forward_norm) + 1),
    "source_dataset_id": 1,
    "player_id": forward_norm["player_id"],
    "club_id": forward_norm["club_id"],
    "original_player_name": forward_norm["nombre"],
    "original_team_name": forward_norm["team"],
    "player_age_years": forward_norm["age"],
    "market_value_mln": forward_norm["value"],
    "value_rank": forward_norm["value_r"],
    "plays_in_europe": forward_norm["europa"],
    "matches_played": forward_norm["matches"],
    "goals": forward_norm["goals"],
    "assists": forward_norm["assists"],
    "minutes_per_goal": forward_norm["mpg"],
    "minutes_played": forward_norm["minutes"],
    "instagram_followers_mln": forward_norm["insta"],
})

forward_fact["instagram_followers_mln"] = (
    forward_fact["instagram_followers_mln"]
    .fillna(0.0)
)
```

```

transfer_norm = transfer_df.copy()

transfer_norm = transfer_norm.merge(
    player_df,
    on="player_name",
    how="left",
)

transfer_norm = transfer_norm.merge(
    club_df,
    on="club_name",
    how="left",
)

transfer_norm = transfer_norm.merge(
    position_df,
    on="position",
    how="left",
)

transfer_norm = transfer_norm.merge(
    nationality_df,
    on="nationality",
    how="left",
)

transfer_fact = pd.DataFrame({
    "transfer_observation_id": range(1, len(transfer_norm) + 1),
    "source_dataset_id": 2,
    "player_id": transfer_norm["player_id"],
    "position_id": transfer_norm["position_id"],
    "nationality_id": transfer_norm["nationality_id"],
    "club_id": transfer_norm["club_id"],
    "season_year": transfer_norm["season"],
    "original_player_name": transfer_norm["player_name"],
    "original_position_name": transfer_norm["position"],
    "original_nationality_name": transfer_norm["nationality"],
    "original_club_name": transfer_norm["club_name"],
    "age_then_years": transfer_norm["age_then"],
    "age_now_years": transfer_norm["age_now"],
    "club_performance": transfer_norm["club_performance"],
    "relegation": transfer_norm["relegation"],
    "success_or_not": transfer_norm["success_or_not"],
    "total_games": transfer_norm["total_games"],
    "assists": transfer_norm["assists"],
    "penalty_kicks": transfer_norm["penalty_kicks"],
    "total_minutes": transfer_norm["total_minutes"],
    "total_goals": transfer_norm["total_goals"],
    "height_cm": transfer_norm["height"],
    "start_value_eur": transfer_norm["start_value"],
    "end_value_eur": transfer_norm["end_value"],
    "delta_value_eur": transfer_norm["delta_value"],
    "value_start_mln": transfer_norm["value_0_mln"],
    "value_end_mln": transfer_norm["value_end_mln"],
    "value_delta_mln": transfer_norm["value_delta_mln"],
})

transfer_fact["club_performance"] = (
    transfer_fact["club_performance"]
    .fillna("")
)

transfer_fact["relegation"] = (
    transfer_fact["relegation"]
    .fillna("")
)

transfer_fact["success_or_not"] = (
    transfer_fact["success_or_not"]
    .fillna("")
)

transfer_fact["start_value_eur"] = (
    transfer_fact["start_value_eur"]
    .fillna(0.0)
)

transfer_fact["end_value_eur"] = (
    transfer_fact["end_value_eur"]
    .fillna(0.0)
)

transfer_fact["delta_value_eur"] = (
    transfer_fact["delta_value_eur"]
    .fillna(0.0)
)

transfer_fact["relegation"] = transfer_fact["relegation"].replace({
    1.0: 1,
    0.0: 0,
})
```

```

tables = {
    "player.csv": player_df,
    "club.csv": club_df,
    "position.csv": position_df,
    "nationality.csv": nationality_df,
    "source_dataset.csv": source_dataset_df,
    "forward_player_valuation.csv": forward_fact,
    "transfer_value_observation.csv": transfer_fact,
}

for filename, df in tables.items():

    path = EXPORT_DIR / filename

    df.to_csv(
        path,
        index=False,
        encoding="utf-8-sig",
    )

    print(f"Exported: {path}")

print("All CSV files exported.")
```
