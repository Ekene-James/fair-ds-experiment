# DBRepo Normalisation + CSV Export
**Project:** Predicting the Market Value of Football Players  
**Purpose:** Normalise both raw Excel datasets into 3NF tables matching the DBRepo schema,
then export clean CSVs ready for manual upload via the DBRepo UI.

## Issues fixed in this version
| Issue | Table | Fix |
|-------|-------|-----|
| `instagram_followers_mln` has 6 NaN | `forward_player_valuation` | Filled with `0.0` (player has no public Instagram) |
| `club_performance`, `relegation`, `success_or_not` are NaN for 2020–2023 | `transfer_value_observation` | Exported as empty string — DBRepo UI treats blank CSV cell as NULL |
| `start_value_eur` has 17 NaN, `end_value_eur` has 2 NaN | `transfer_value_observation` | Same — exported as empty string for NULL |
| `plays_in_europe` stored as int 0/1 | `forward_player_valuation` | Converted to `0`/`1` integer — DBRepo BOOLEAN accepts this |
| One garbled player name `脙聡aglar Soyuncu` | `player`, `transfer_value_observation` | Corrected to `caglar Soyuncu` (matches other rows for same player) |
| Special characters in names (accents, diacritics) | All name columns | CSV exported with UTF-8 BOM (`utf-8-sig`) so the DBRepo UI parser reads them correctly |
| `season` table was missing | `season` | Added with correct 5 rows matching the schema SQL |
| `source_dataset` was incomplete | `source_dataset` | All 11 schema columns now populated |
| `position` column named `position` not `position_name` | `position` | Renamed to match schema |
| `nationality` column named `nationality` not `nationality_name` | `nationality` | Renamed to match schema |

## Upload order in the DBRepo UI
You must upload in this order to respect foreign key constraints:
```
1. source_dataset
2. player
3. club
4. position
5. nationality
6. season
7. forward_player_valuation
8. transfer_value_observation
```

## 1. Imports

````python
import pandas as pd
import numpy as np
from pathlib import Path
````

## 2. Configuration
Update `DATA_DIR` to the folder containing your two Excel files.

````python
DATA_DIR      = Path("../data")   # <-- update if needed
FORWARD_FILE  = DATA_DIR / "Dataset soccerplayers.xlsx"
TRANSFER_FILE = DATA_DIR / "Nisanov_final_data.xlsx"

EXPORT_DIR = Path("normalized_exports")
EXPORT_DIR.mkdir(exist_ok=True)

print(f"Forward file  exists: {FORWARD_FILE.exists()}")
print(f"Transfer file exists: {TRANSFER_FILE.exists()}")
print(f"Export dir          : {EXPORT_DIR.resolve()}")
````

## 3. Load Raw Data

````python
# ── Dataset 1 ─────────────────────────────────────────────────────────────────
forward_raw = pd.read_excel(FORWARD_FILE)
forward_raw.columns = [c.strip().lower() for c in forward_raw.columns]
print(f"Dataset 1: {forward_raw.shape}  columns: {forward_raw.columns.tolist()}")

# ── Dataset 2 (5 sheets: 2019–2023) ───────────────────────────────────────────
# Sheets 2020–2023 lack club_performance/relegation/success_or_not.
# After concat those columns are NaN for those rows — correct, schema allows NULL.
transfer_sheets = pd.read_excel(TRANSFER_FILE, sheet_name=None)
transfer_raw = pd.concat(
    [sheet.dropna(axis=1, how="all") for sheet in transfer_sheets.values()],
    ignore_index=True,
    sort=False,
)
transfer_raw.columns = [c.strip().lower() for c in transfer_raw.columns]
print(f"Dataset 2: {transfer_raw.shape}  columns: {transfer_raw.columns.tolist()}")
````

## 4. Fix Known Data Quality Issues

````python
# ── Fix 1: Garbled player name in Dataset 2 ───────────────────────────────────
# '脙聡aglar Soyuncu' is a UTF-8 encoding corruption of the Turkish name.
# All other rows for this player use 'caglar Soyuncu' so we normalise to that.
garbled = transfer_raw['player_name'] == '脙聡aglar Soyuncu'
print(f"Garbled name rows found: {garbled.sum()}")
transfer_raw.loc[garbled, 'player_name'] = 'caglar Soyuncu'
print("Fixed: '脙聡aglar Soyuncu' → 'caglar Soyuncu'")

# ── Fix 2: instagram_followers_mln — 6 NaN rows ───────────────────────────────
# These players have no public Instagram account.
# Schema column is DECIMAL(12,3) NOT NULL is not declared, so NULL is fine,
# but the DBRepo UI rejects blank cells in non-nullable numeric columns.
# We fill with 0.0 which is semantically correct (0 million followers).
insta_nulls = forward_raw['insta'].isna().sum()
print(f"\nInstagram NaN rows: {insta_nulls}")
forward_raw['insta'] = forward_raw['insta'].fillna(0.0)
print("Fixed: instagram NaN → 0.0")

# ── Fix 3: Nullable columns in transfer — leave as NaN ────────────────────────
# club_performance (1157 NaN), relegation (1157), success_or_not (1139)
# start_value_eur (17 NaN), end_value_eur (2 NaN)
# These are genuinely NULL in the source data (only 2019 sheet has them).
# We do NOT fill them — the export helper will write empty cells for NaN,
# which the DBRepo UI correctly interprets as NULL.
print("\nNullable column NaN counts (will be exported as empty cells = NULL):")
nullable_cols = ['club_performance','relegation','success_or_not','start_value','end_value']
for col in nullable_cols:
    print(f"  {col}: {transfer_raw[col].isna().sum()} NaN")

print("\nData quality fixes complete.")
````

## 5. Build Lookup Tables

````python
# ── player ────────────────────────────────────────────────────────────────────
player_names = pd.concat([
    forward_raw['nombre'].rename('player_name'),
    transfer_raw['player_name'],
]).dropna().drop_duplicates().sort_values().reset_index(drop=True)

player_df = pd.DataFrame({
    'player_id':   range(1, len(player_names) + 1),
    'player_name': player_names.values,
})
print(f"player:      {len(player_df):>5} rows")

# ── club ──────────────────────────────────────────────────────────────────────
club_names = pd.concat([
    forward_raw['team'].rename('club_name'),
    transfer_raw['club_name'],
]).dropna().drop_duplicates().sort_values().reset_index(drop=True)

club_df = pd.DataFrame({
    'club_id':   range(1, len(club_names) + 1),
    'club_name': club_names.values,
})
print(f"club:        {len(club_df):>5} rows")

# ── position ──────────────────────────────────────────────────────────────────
# Column MUST be named 'position_name' to match schema — not 'position'
position_names = (
    transfer_raw['position']
    .dropna().drop_duplicates().sort_values().reset_index(drop=True)
)
position_df = pd.DataFrame({
    'position_id':   range(1, len(position_names) + 1),
    'position_name': position_names.values,
})
print(f"position:    {len(position_df):>5} rows")

# ── nationality ───────────────────────────────────────────────────────────────
# Column MUST be named 'nationality_name' to match schema — not 'nationality'
nationality_names = (
    transfer_raw['nationality']
    .dropna().drop_duplicates().sort_values().reset_index(drop=True)
)
nationality_df = pd.DataFrame({
    'nationality_id':   range(1, len(nationality_names) + 1),
    'nationality_name': nationality_names.values,
})
print(f"nationality: {len(nationality_df):>5} rows")

# ── season ────────────────────────────────────────────────────────────────────
# Matches the INSERT statements in schema.sql exactly
season_df = pd.DataFrame({
    'season_year': [2019, 2020, 2021, 2022, 2023],
    'notes': ['Season sheet in Transfer Value Determinants'] * 5,
})
print(f"season:      {len(season_df):>5} rows")
````

## 6. Build `source_dataset` Table
All 11 schema columns must be present.

````python
source_dataset_df = pd.DataFrame([
    {
        'source_dataset_id': 1,
        'dataset_title':     'Forward football player valuation',
        'original_creators': 'Hugo Briseño; José Carlos Rivera',
        'publisher':         'Mendeley Data',
        'version_label':     'V1',
        'doi':               '10.17632/cgc33scxg7.1',
        'doi_url':           'https://doi.org/10.17632/cgc33scxg7.1',
        'license_name':      'CC BY 4.0',
        'license_url':       'https://creativecommons.org/licenses/by/4.0/',
        'source_filename':   'soccerplayers.xlsx',
        'notes':             'Reused external source dataset. The group republishes/uses the data but is not the original publisher or rights holder.',
    },
    {
        'source_dataset_id': 2,
        'dataset_title':     'Transfer Value Determinants',
        'original_creators': 'Ronald Nisanov',
        'publisher':         'Mendeley Data',
        'version_label':     'V2',
        'doi':               '10.17632/3btg6ptc7b.2',
        'doi_url':           'https://doi.org/10.17632/3btg6ptc7b.2',
        'license_name':      'CC BY 4.0',
        'license_url':       'https://creativecommons.org/licenses/by/4.0/',
        'source_filename':   'Nisanov_final_data.xlsx',
        'notes':             'Reused external source dataset. The group republishes/uses the data but is not the original publisher or rights holder.',
    },
])
print(f"source_dataset: {len(source_dataset_df)} rows")
print(f"Columns: {source_dataset_df.columns.tolist()}")
````

## 7. Build `forward_player_valuation` Fact Table

````python
fwd = forward_raw.copy()

# Join player_id
fwd = fwd.merge(
    player_df.rename(columns={'player_name': 'nombre'}),
    on='nombre', how='left',
)
# Join club_id
fwd = fwd.merge(
    club_df.rename(columns={'club_name': 'team'}),
    on='team', how='left',
)

# Verify all rows matched
assert fwd['player_id'].isna().sum() == 0, "Unmatched players in forward dataset"
assert fwd['club_id'].isna().sum() == 0,   "Unmatched clubs in forward dataset"

forward_fact = pd.DataFrame({
    'forward_valuation_id':    range(1, len(fwd) + 1),
    'source_dataset_id':       1,
    'player_id':               fwd['player_id'].astype(int),
    'club_id':                 fwd['club_id'].astype(int),
    'original_player_name':    fwd['nombre'],
    'original_team_name':      fwd['team'],
    'player_age_years':        fwd['age'].astype(int),
    'market_value_mln':        fwd['value'],
    'value_rank':              fwd['value_r'].astype(int),
    # BOOLEAN: export as 0/1 integer — DBRepo UI accepts this for BOOLEAN columns
    'plays_in_europe':         fwd['europa'].astype(int),
    'matches_played':          fwd['matches'].astype(int),
    'goals':                   fwd['goals'].astype(int),
    'assists':                 fwd['assists'].astype(int),
    'minutes_per_goal':        fwd['mpg'].astype(int),
    'minutes_played':          fwd['minutes'].astype(int),
    # NaN filled with 0.0 in Section 4 — no nulls remain
    'instagram_followers_mln': fwd['insta'],
})

# Final null check — must all be 0
nulls = forward_fact.isnull().sum()
remaining_nulls = nulls[nulls > 0]
if len(remaining_nulls) == 0:
    print(f"forward_player_valuation: {len(forward_fact)} rows — no nulls")
else:
    print(f"WARNING: nulls remain:\n{remaining_nulls}")
````

## 8. Build `transfer_value_observation` Fact Table

````python
trf = transfer_raw.copy()

# Join all lookup IDs
trf = trf.merge(player_df, on='player_name', how='left')
trf = trf.merge(club_df,   on='club_name',   how='left')
trf = trf.merge(
    position_df.rename(columns={'position_name': 'position'}),
    on='position', how='left',
)
trf = trf.merge(
    nationality_df.rename(columns={'nationality_name': 'nationality'}),
    on='nationality', how='left',
)

# Verify all FK rows matched
assert trf['player_id'].isna().sum() == 0,      "Unmatched players"
assert trf['club_id'].isna().sum() == 0,         "Unmatched clubs"
assert trf['position_id'].isna().sum() == 0,     "Unmatched positions"
assert trf['nationality_id'].isna().sum() == 0,  "Unmatched nationalities"
print("All FK lookups matched.")

transfer_fact = pd.DataFrame({
    'transfer_observation_id':   range(1, len(trf) + 1),
    'source_dataset_id':         2,
    'player_id':                 trf['player_id'].astype(int),
    'position_id':               trf['position_id'].astype(int),
    'nationality_id':            trf['nationality_id'].astype(int),
    'club_id':                   trf['club_id'].astype(int),
    'season_year':               trf['season'].astype(int),
    'original_player_name':      trf['player_name'],
    'original_position_name':    trf['position'],
    'original_nationality_name': trf['nationality'],
    'original_club_name':        trf['club_name'],
    'age_then_years':            trf['age_then'].astype(int),
    'age_now_years':             trf['age_now'].astype(int),
    # These 3 are NULL for 2020-2023 rows — kept as NaN, exported as empty cell
    'club_performance':          trf['club_performance'],   # float, NaN OK
    'relegation':                trf['relegation'],         # float, NaN OK
    'success_or_not':            trf['success_or_not'],     # float, NaN OK
    'total_games':               trf['total_games'].astype(int),
    'assists':                   trf['assists'].astype(int),
    'penalty_kicks':             trf['penalty_kicks'].astype(int),
    'total_minutes':             trf['total_minutes'].astype(int),
    'total_goals':               trf['total_goals'].astype(int),
    'height_cm':                 trf['height'].astype(float),
    # start_value_eur: 17 NaN, end_value_eur: 2 NaN — exported as empty cell
    'start_value_eur':           trf['start_value'],
    'end_value_eur':             trf['end_value'],
    'delta_value_eur':           trf['delta_value'].astype(float),
    'value_start_mln':           trf['value_0_mln'],
    'value_end_mln':             trf['value_end_mln'],
    'value_delta_mln':           trf['value_delta_mln'],
})

# Show null summary — only the known nullable columns should appear here
nulls = transfer_fact.isnull().sum()
remaining = nulls[nulls > 0]
print(f"\ntransfer_value_observation: {len(transfer_fact)} rows")
print("Null counts (these will be empty cells in CSV = NULL in DB):")
print(remaining.to_string())
````

## 9. Export Clean CSVs

**Key export settings:**
- `encoding='utf-8-sig'` — UTF-8 with BOM. This ensures the DBRepo UI correctly 
  reads accented characters (é, ñ, ı, etc.) in player and club names.
- `na_rep=''` — NaN values become empty cells in the CSV, which the DBRepo UI 
  correctly treats as NULL for nullable columns.
- `index=False` — no row numbers in the output.

````python
def export_csv(df, filename, export_dir):
    """Export a DataFrame to CSV with settings optimised for DBRepo UI upload."""
    path = export_dir / filename
    df.to_csv(
        path,
        index=False,
        encoding='utf-8-sig',  # BOM ensures correct special character handling
        na_rep='',             # NaN → empty cell → NULL in DBRepo
    )
    size_kb = path.stat().st_size / 1024
    print(f"  {filename:<45} {len(df):>5} rows  {size_kb:>7.1f} KB")


# Export in upload order
print("Exporting CSVs...")
print(f"{'File':<45} {'Rows':>5}  {'Size':>8}")
print("-" * 62)

export_csv(source_dataset_df, 'source_dataset.csv',             EXPORT_DIR)
export_csv(player_df,         'player.csv',                     EXPORT_DIR)
export_csv(club_df,           'club.csv',                       EXPORT_DIR)
export_csv(position_df,       'position.csv',                   EXPORT_DIR)
export_csv(nationality_df,    'nationality.csv',                EXPORT_DIR)
export_csv(season_df,         'season.csv',                     EXPORT_DIR)
export_csv(forward_fact,      'forward_player_valuation.csv',   EXPORT_DIR)
export_csv(transfer_fact,     'transfer_value_observation.csv', EXPORT_DIR)

print("-" * 62)
print(f"All CSVs saved to: {EXPORT_DIR.resolve()}")
````

## 10. Final Verification
Sanity-check each exported CSV before uploading.

````python
print("=" * 60)
print("FINAL VERIFICATION")
print("=" * 60)

checks = [
    ('source_dataset.csv',             source_dataset_df, 2),
    ('player.csv',                     player_df,         None),
    ('club.csv',                       club_df,           None),
    ('position.csv',                   position_df,       None),
    ('nationality.csv',                nationality_df,    None),
    ('season.csv',                     season_df,         5),
    ('forward_player_valuation.csv',   forward_fact,      438),
    ('transfer_value_observation.csv', transfer_fact,     2502),
]

all_ok = True
for filename, df, expected_rows in checks:
    path = EXPORT_DIR / filename
    # Re-read the exported CSV to confirm it round-trips correctly
    reloaded = pd.read_csv(path, encoding='utf-8-sig')
    row_ok  = (expected_rows is None) or (len(reloaded) == expected_rows)
    col_ok  = list(reloaded.columns) == list(df.columns)
    status  = "OK" if (row_ok and col_ok) else "FAIL"
    if status == "FAIL":
        all_ok = False
    print(f"[{status}] {filename}")
    if not col_ok:
        print(f"       Column mismatch!")
        print(f"       Expected: {list(df.columns)}")
        print(f"       Got:      {list(reloaded.columns)}")
    if not row_ok:
        print(f"       Row count mismatch: expected {expected_rows}, got {len(reloaded)}")

print()
if all_ok:
    print("All checks passed. CSVs are ready for upload via the DBRepo UI.")
    print()
    print("Upload order in DBRepo UI:")
    for i, (f, _, _) in enumerate(checks, 1):
        print(f"  {i}. {f}")
else:
    print("Some checks failed — fix before uploading.")
````
